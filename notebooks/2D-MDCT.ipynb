### Basic example

```python
%%bash
rm -f /tmp/encoded*
python ../src/2D-MDCT.py encode
rm -f /tmp/decoded.png
python ../src/2D-MDCT.py decode
python ../src/RDE.py
```

### Using perceptual (low) quantization

```python
%%bash
rm -f /tmp/encoded*
python ../src/2D-MDCT.py encode -p -q 4 -B 32
rm -f /tmp/decoded.png
python ../src/2D-MDCT.py decode -p -q 4 -B 32
python ../src/RDE.py
```

```python
display(Image(filename="/tmp/decoded.png"))
```

### Using perceptual (high) quantization

```python
%%bash
rm -f /tmp/encoded*
python ../src/2D-MDCT.py encode -p -q 8 -B 32
rm -f /tmp/decoded.png
python ../src/2D-MDCT.py decode -p -q 8 -B 32
python ../src/RDE.py
```

```python
display(Image(filename="/tmp/decoded.png"))
```

### Using the LloydMax quantizer

```python
%%bash
rm -f /tmp/encoded*
python ../src/2D-MDCT.py encode -a LloydMax -m -2048 -n 2047
rm -f /tmp/decoded.png
python ../src/2D-MDCT.py decode -a LloydMax -m -2048 -n 2047
python ../src/RDE.py
```

```python
display(Image(filename="/tmp/decoded.png"))
```
